##Data cleaning

###Finding duplicates

```python
df.duplicated().sum()
```

```python
df['author'].nunique()
```

###Clean `author` and `narrator` columns

```python
df['author'] = df['author'].str.replace('Writtenby:', '', regex=False)
df['narrator'] = df['narrator'].str.replace('Narratedby:', '', regex=False)
df.sample(10)
```

Clean `price` and `language` column

```python
df['price'] = df['price'].str.replace(r'\..*', '', regex=True)
df['language'] = df['language'].str.capitalize()
df.sample(10)
```

###Clean `author` and 'narrator`

```python
df['author'] = df['author'].str.replace(r'(?<=[a-z,])(?=[A-Z])', ' ', regex = True)
df['narrator'] = df['narrator'].str.replace(r'(?<=[a-z,])(?=[A-Z])', ' ', regex = True)
df.sample(10)
```

###Change the type of `relesedate`

```python
df['releasedate'] = pd.to_datetime(df['releasedate'], errors = 'coerce')
df.head()
```

```python
df.info()
```

###Working with `time` column

```python
df['time'] = df['time'].str.replace(r'[a-zA-Z]', '', regex = True)
df['time'] = df['time'].str.strip().str.split()
df.head()
```

```python
def convert_time(x):
    if len(x) >= 2:
      return int(x[0])*60 + int(x[1])
    elif len(x) < 2:
      return int(x[0])
    return None

df['time'] = df['time'].apply(convert_time)
df.sample(10)
```

###Cleaning foreign names

```python
mask = ~df['name'].str.contains(r'^[A-Za-z0-9]', regex = True, na = False)
df = df.drop(index = df[mask].index)
df.sample(10)
```

###Working with `stars` column

```python
df['stars'] = df['stars'].str.replace(r'[A-Za-z]', '', regex = True).str.strip().str.split()
df.sample(10)
```

```python
def convert_star(x):
  if len(x) >=2:
    return float(x[0])
  elif len(x) < 2:
    return None
  return None

df['stars'] = df['stars'].apply(convert_star)
df.sample(10)
```

```python
df['stars'] = df['stars'].fillna(df['stars'].mean().round(2))
```

```python
df.sample(10)
```

Convert dataframe to csv

```python
df.to_csv('cleaned_data1.csv', index = False)
```
